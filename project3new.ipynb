```python
import numpy as np

electronics = monthly_category_sales[['Electronics']].copy()

electronics['t'] = np.arange(1, len(electronics) + 1)

electronics
```

```python
from sklearn.linear_model import LinearRegression

X = electronics[['t']]
y = electronics['Electronics']

model = LinearRegression()
model.fit(X, y)
```

```python
a = model.intercept_
b = model.coef_[0]

print("Intercept (a):", a)
print("Slope (b):", b)
print(f"Regression Equation: Sales = {a:.2f} + {b:.2f}t")
```

```python
r_squared = model.score(X, y)

print("R²:", r_squared)
```

```python
future_t = np.arange(len(electronics) + 1, len(electronics) + 7)

future_sales = model.predict(future_t.reshape(-1, 1))

future_forecast = pd.DataFrame({
    't': future_t,
    'Forecast_Sales': future_sales
})

future_forecast
```

```python
latest_actual = electronics['Electronics'].iloc[-1]
forecast_last = future_forecast['Forecast_Sales'].iloc[-1]

forecast_growth = ((forecast_last - latest_actual) / latest_actual) * 100

print("Latest Actual Sales:", latest_actual)
print("Forecast Sales:", forecast_last)
print("Forecast Growth %:", forecast_growth)
```

```python
latest_actual = electronics['Electronics'].iloc[-1]
forecast_3_month = future_forecast['Forecast_Sales'].iloc[2]

forecast_growth = ((forecast_3_month - latest_actual) / latest_actual) * 100

print("Latest Actual Sales:", latest_actual)
print("3-Month Forecast Sales:", forecast_3_month)
print("3-Month Forecast Growth %:", forecast_growth)
```
